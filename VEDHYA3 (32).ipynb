```python
%%writefile static/script.js
console.log("✅ Vedhya script loaded!");

fetch("/data")
    .then(response => {
        if (!response.ok) throw new Error(`HTTP ${response.status}`);
        return response.json();
    })
    .then(data => {
        console.log("✅ Data loaded:", data);
        const list = document.getElementById("dataList");
        if (!list) {
            console.error("❌ #dataList element missing");
            return;
        }
        // ✅ Pretty formatting instead of JSON.stringify
        data.forEach(item => {
            const li = document.createElement("li");
            li.className = "list-group-item bg-transparent border-glass";
            li.innerHTML = `<strong>${item.name || 'Item'}</strong><br><small>${item.value || 'N/A'}</small>`;
            list.appendChild(li);
        });
    })
    .catch(error => {
        console.error("❌ Fetch failed:", error);
        // ✅ FIXED: Consistent ID + better mock data
        const mockData = [
            {"id": 1, "name": "Python Basics", "value": "Completed"},
            {"id": 2, "name": "AI Chatbot", "value": "Active"}
        ];
        const list = document.getElementById("dataList");
        if (list) {
            mockData.forEach(item => {
                const li = document.createElement("li");
                li.className = "list-group-item bg-transparent border-glass";
                li.innerHTML = `<strong>${item.name}</strong>: ${item.value} <span class="badge bg-success ms-2">(Demo)</span>`;
                list.appendChild(li);
            });
            console.log("✅ Mock data loaded");
        }
    });
```

```python
import os
from pymongo import MongoClient
from dotenv import load_dotenv
from datasets import load_dataset
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    Trainer,
    TrainingArguments,
    DataCollatorForLanguageModeling
)

def train_vedhya_ai():
    load_dotenv()

    # 1. System Configuration
    model_name = "microsoft/DialoGPT-small"
    train_path = "/content/data/train.txt"
    output_dir = "./vedhya-model"

    # 2. Database Sync (Fixing the collection name typo)
    if not os.path.exists("/content/data"): os.makedirs("/content/data")

    try:
        client = MongoClient(os.getenv('MONGO_URI'))
        db = client[os.getenv('DATABASE_NAME', 'project')]
        intents_col = db['training_ddta'] # Ensure this matches Cluster0

        with open(train_path, "w") as f:
            for item in intents_col.find({}):
                for pattern in item.get("patterns", []): f.write(pattern + "\n")
                for response in item.get("responses", []): f.write(response + "\n")
        print("✅ MongoDB Intelligence Core Synced.")
    except Exception as e:
        print(f"❌ DB Sync Failed: {e}")
        return

    # 3. Modern Neural Processing
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    tokenizer.pad_token = tokenizer.eos_token # Critical for CausalLM

    def tokenize_function(examples):
        return tokenizer(examples["text"], truncation=True, padding="max_length", max_length=64)

    # Load dataset using the modern library
    dataset = load_dataset("text", data_files={"train": train_path})
    tokenized_dataset = dataset.map(tokenize_function, batched=True, remove_columns=["text"])

    data_collator = DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False)

    # 4. Professional Training Parameters
    training_args = TrainingArguments(
        output_dir=output_dir,
        num_train_epochs=5,
        per_device_train_batch_size=4,
        learning_rate=5e-5,
        save_total_limit=1,
        save_strategy="epoch", # Added save_strategy
        save_steps=500, # Added save_steps
        report_to="none",
        logging_dir="./logs" # Added logging_dir
    )

    # 5. Initialize VEDHYA Sync
    trainer = Trainer(
        model=AutoModelForCausalLM.from_pretrained(model_name),
        args=training_args,
        train_dataset=tokenized_dataset["train"],
        data_collator=data_collator,
    )

    print("🚀 Initializing VEDHYA Neural Training...")
    trainer.train()

    # 6. Finalize Save
    trainer.save_model(output_dir)
    tokenizer.save_pretrained(output_dir)
    print(f"✅ VEDHYA OS Brain updated at {output_dir}")

if __name__ == "__main__":
    train_vedhya_ai()
```

```python
%%writefile app.py
import os
import random
import re
import threading
from datetime import datetime
from dotenv import load_dotenv
from flask import Flask, render_template, request, session, redirect, url_for, flash
from flask_socketio import SocketIO, emit
from flask_wtf.csrf import CSRFProtect
from flask_mail import Mail, Message
from itsdangerous import URLSafeTimedSerializer, SignatureExpired
from functools import wraps
from pymongo import MongoClient

load_dotenv()

app = Flask(__name__)
app.secret_key = os.getenv('FLASK_SECRET_KEY', 'vedhya-ai-secure-2026')

# --- MONGODB CLUSTER0 INTEGRATION ---
client = MongoClient(os.getenv('MONGO_URI'))
db = client[os.getenv('DATABASE_NAME', 'project')]
intents_col = db['training_ddta']  # Fixed typo
users_col = db['users']

# --- SYSTEM CONFIGURATION ---
app.config.update(
    MAIL_SERVER='smtp.gmail.com',
    MAIL_PORT=587,
    MAIL_USE_TLS=True,
    MAIL_USERNAME=os.environ.get('MAIL_USERNAME'),
    MAIL_PASSWORD=os.environ.get('MAIL_PASSWORD'),
    MAIL_DEFAULT_SENDER=os.environ.get('MAIL_USERNAME')
)

mail = Mail(app) if app.config['MAIL_USERNAME'] else None
csrf = CSRFProtect(app)
socketio = SocketIO(app, cors_allowed_origins="*", async_mode='threading')
serializer = URLSafeTimedSerializer(app.secret_key)

# --- CONTINUOUS TELEMETRY STREAM ---
def watch_database_changes():
    """ Monitors MongoDB for updates and pushes live telemetry to the UI. """
    try:
        # Utilizing Change Streams for real-time reactivity
        with users_col.watch(full_document='updateLookup') as stream:
            for change in stream:
                if change['operationType'] == 'update':
                    doc = change['fullDocument']
                    # Push only relevant engineering telemetry
                    socketio.emit('telemetry_update', {
                        "xp": doc.get('xp', 2840),
                        "velocity": [random.randint(100, 900) for _ in range(7)]
                    })
    except Exception as e:
        print(f"⚠️ Telemetry Stream Offline: {e}")

# --- SECURITY DECORATOR ---
def verification_required(f):
    @wraps(f)
    def decorated_function(*args, **kwargs):
        if not session.get('username'):
            return redirect(url_for('login'))
        if not session.get('is_verified'):
            flash('ACCESS_RESTRICTED: Terminal locked. Verify email to unlock AI modules.', 'warning')
            return redirect(url_for('home'))
        return f(*args, **kwargs)
    return decorated_function

# --- INTENT ENGINE ---
def get_ai_response(user_input):
    clean_input = re.sub(r'[^\w\s]', '', user_input.lower().strip())
    match = intents_col.find_one({
        "patterns": {"$regex": re.escape(clean_input), "$options": "i"}
    })
    if match:
        return random.choice(match['responses'])
    return "<b>QUERY_UNRESOLVED:</b> Intent not found in VEDHYA Core. Try asking about 'Python' or 'MERN'."

# --- CORE ROUTES ---

@app.route('/')
def home():
    if session.get('is_verified'):
        return redirect(url_for('dashboard'))
    return render_template('home.html')

@app.route('/about')
def about():
    return render_template('about.html')

@app.route('/contact', methods=['GET', 'POST'])
def contact():
    if request.method == 'POST':
        name = request.form.get('name')
        user_email = request.form.get('email')
        msg_body = request.form.get('message')
        if mail:
            try:
                admin_msg = Message(
                    subject=f"New Inquiry: {name}",
                    recipients=[app.config['MAIL_USERNAME']],
                    body=f"From: {name} ({user_email})\n\nMessage: {msg_body}"
                )
                mail.send(admin_msg)
                flash('SYNC_SUCCESS: Admin has been notified.', 'success')
            except Exception:
                flash('MAIL_ERROR: System offline.', 'danger')
        return redirect(url_for('home'))
    return render_template('contact.html')

@app.route('/register', methods=['GET', 'POST'])
def register():
    if request.method == 'POST':
        email = request.form.get('email', '').strip().lower()
        password = request.form.get('password')
        existing_user = users_col.find_one({"email": email})
        if existing_user:
            flash('ACCOUNT_EXISTS: This email is already registered.', 'warning')
            return redirect(url_for('login'))
        if mail:
            try:
                token = serializer.dumps(email, salt='email-confirm')
                confirm_url = url_for('confirm_email', token=token, _external=True)
                msg = Message('Initialize VEDHYA OS Terminal', recipients=[email])
                msg.body = f'Welcome, operative. Activate your terminal access here: {confirm_url}'
                mail.send(msg)
                users_col.insert_one({
                    "email": email,
                    "password": password,
                    "is_verified": False,
                    "created_at": datetime.now()
                })
                flash('VERIFICATION_SENT: Check your email to activate account.', 'info')
                return redirect(url_for('login'))
            except Exception:
                flash('MAIL_ERROR: Communication link down.', 'danger')
    return render_template('register.html')

@app.route('/confirm/<token>')
def confirm_email(token):
    try:
        email = serializer.loads(token, salt='email-confirm', max_age=3600)
        users_col.update_one({"email": email}, {"$set": {"is_verified": True}})
        session['username'] = email.split('@')[0]
        session['is_verified'] = True
        flash('TERMINAL_ACTIVATED: Authorization confirmed.', 'success')
        return redirect(url_for('dashboard'))
    except (SignatureExpired, Exception):
        flash('ERROR: Activation link is invalid or expired.', 'danger')
        return redirect(url_for('login'))

@app.route('/login', methods=['GET', 'POST'])
def login():
    if request.method == 'POST':
        email = request.form.get('email', '').strip().lower()
        user = users_col.find_one({"email": email})
        if user:
            session['username'] = email.split('@')[0]
            session['is_verified'] = user.get('is_verified', False)
            return redirect(url_for('dashboard'))
        flash('AUTH_FAILED: Credentials not recognized.', 'danger')
    return render_template('login.html')

@app.route('/dashboard')
@verification_required
def dashboard():
    # Pull current telemetry from Cluster0
    user_email = session.get('username') + "@gmail.com"
    user_data = users_col.find_one({"email": user_email}) or {}
    stats = {
        "xp": user_data.get('xp', 2840),
        "streak": user_data.get('streak', 7),
        "completion_rate": 94,
        "active_courses": 4
    }
    return render_template('dashboard.html', stats=stats)

@app.route('/chatbot')
@verification_required
def chatbot():
    return render_template('chatbot.html')

@app.route('/logout')
def logout():
    session.clear()
    return redirect(url_for('home'))

@socketio.on('connect')
def handle_connect():
    """ Initialize background watch thread on connection. """
    global thread
    if 'thread' not in globals():
        socketio.start_background_task(watch_database_changes)

@socketio.on('message')
def handle_message(data):
    if not session.get('is_verified'): return
    user_msg = data.get('msg', '').strip()
    if user_msg:
        emit('new_message', {'user': session.get('username'), 'msg': user_msg}, broadcast=True)
        emit('new_message', {'user': '🤖 Vedhya AI', 'msg': get_ai_response(user_msg)}, broadcast=True)

if __name__ == '__main__':
    socketio.run(app, host='0.0.0.0', port=5000, debug=True, allow_unsafe_werkzeug=True)
```

```python
%%writefile startup.py

import os, subprocess, threading, time, socket
from pyngrok import ngrok
from dotenv import load_dotenv

load_dotenv()

def is_port_listening(port):
    """Checks if the Flask server has successfully bound to the port."""
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        return s.connect_ex(('127.0.0.1', port)) == 0

# 1. Kill any existing processes (Ngrok & Flask)
print("🧹 Cleaning local environment...")
ngrok.kill()
# fuser is more effective in Colab for clearing stuck ports
os.system("fuser -k 5000/tcp || true")
os.system("pkill -f 'python app.py' || true")
time.sleep(3)

# 2. Start Flask Kernel
def start_flask():
    print("🚀 Initializing VEDHYA AI Server...")
    # Use python3 explicitly for Colab's Linux environment
    with open('flask.log', 'w') as out, open('flask_err.log', 'w') as err:
        subprocess.Popen(['python3', 'app.py'], stdout=out, stderr=err)

threading.Thread(target=start_flask, daemon=True).start()

# 3. Wait for stabilization
print("⏳ Waiting for Kernel stabilization (Max 15s)...")
server_online = False
for i in range(15):
    time.sleep(1)
    if is_port_listening(5000):
        print(f"✅ Kernel Online (Sync time: {i+1}s)")
        server_online = True
        break

# 4. Establish External Bridge
if server_online:
    try:
        ngrok.set_auth_token(os.getenv('NGROK_AUTH_TOKEN'))
        public_url = ngrok.connect(5000, bind_tls=True)
        print(f"\n🎉 VEDHYA AI LIVE: {public_url}")

    except Exception as e:
        print(f"❌ Ngrok Tunnel Failed: {e}")
else:
    print("❌ CRITICAL: Flask failed to start. Reviewing flask_err.log...")
    if os.path.exists('flask_err.log'):
        with open('flask_err.log', 'r') as f:
            print(f"\n--- LOG DUMP ---\n{f.read()}\n----------------")

while True: time.sleep(60)
```
